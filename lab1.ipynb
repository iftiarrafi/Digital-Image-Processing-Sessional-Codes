**d) Separate the given color image in three R G & B color channels.**

```python
image1_rgb = cv2.cvtColor(image1 , cv2.COLOR_BGR2RGB)
r,g,b = cv2.split(image1_rgb)
k = np.zeros_like(r)
red_image = cv2.merge([r,k,k])
green_image = cv2.merge([k,g,k])
blue_image = cv2.merge([k,k,b])

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.imshow(red_image)
plt.title("red")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(green_image)
plt.title("green")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blue_image)
plt.title("blue")
plt.axis("off")

plt.show()
```

#Flipping

```python
img = cv2.imread("g.png")
img_rgb = cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
```

```python
hori_flip = cv2.flip(img_rgb , 1)
ver_flip = cv2.flip(img_rgb , 0)
both_flip = cv2.flip(img_rgb , -1)
plt.figure(figsize=(10,5))


plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title("original")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(hori_flip)
plt.title("Horizontal flip")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(ver_flip)
plt.title("Vertical flip")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(both_flip)
plt.title("both flip")
plt.axis("off")

plt.show()
```

**i) Write a program to create a 512X512 full black image and draw line, rectangle,
circle and text on that image.**

```python
h,w=512,512
black_img = np.zeros((h,w,3))
plt.imshow(black_img)
```

```python
# Draw a green diagonal line from top-left to bottom-right

cv2.line(black_img , (0,0) , (512,512) , (0,255,0) , 2)
plt.imshow(black_img)
```

```python
# Draw a filled blue rectangle (top-left and bottom-right coordinates)
cv2.rectangle(black_img , (100,100) , (200,200) , (0,0,255) , -1)
plt.imshow(black_img)
```

```python
# Draw a filled red circle (center coordinates and radius)
c_w , c_h = int(black_img.shape[0] * 0.5) , int(black_img.shape[1] * 0.5)
radius = 50
cv2.circle(black_img , (c_w , c_h) , radius , (255 , 0 , 0) , -1)
plt.imshow(black_img)
```

```python
# Add text to the image
cv2.putText(
    img = black_img,
    text= "Hello Rafi 2",
    org = (200,200),
    fontFace = cv2.FONT_HERSHEY_TRIPLEX,
    fontScale = 1,
    color = (122,133,566),
    thickness = 1,
)

plt.imshow(black_img)
```

# --- 1. SPATIAL RESOLUTION REDUCTION ---

```python
img = cv2.imread("g.png")
img_rgb = cv2.cvtColor(img , cv2.COLOR_BGR2RGB) 
```

```python
downsampled_images = []
n_values=[200,100,50,20]

for n in n_values:
    downed_img = cv2.resize(img_rgb , (n,n) , interpolation=cv2.INTER_AREA)
    downsampled_images.append(downed_img)

plt.figure(figsize=(10,5))

for i , im in enumerate(downsampled_images) :
    plt.subplot(1,4,i+1)
    plt.imshow(im)
    plt.axis("off")
plt.tight_layout()
plt.show()
```
